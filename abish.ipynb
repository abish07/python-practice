```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import pyplot as plt
import plotly.express as ex
from sklearn.preprocessing import OneHotEncoder,StandardScaler,LabelEncoder
import os
```

```python
fruits = {"Apple":10,"mango":20,"papaya":10,"banana":12,"grapes":50,"watermelon":200,"orange":10,"litchi":20,"Guava":30,"kiwi":4,"avocado":1,"dragon fruits":2}
```

```python
size = fruits.values()
label = fruits.keys()
```

```python
print(label)
```

```python
plt.Figure(figsize=(130,115))
plt.pie(size,labels=label,autopct="%0.01f%%", startangle=140)
plt.title('This is pie chart for the data science')
plt.legend(loc='upper right',bbox_to_anchor=(2,1))
plt.tight_layout()
plt.savefig('piechart.png')
plt.show()
```
